# Experiment II: Data Import, Export and Cleaning (Single Program)

**Dataset:** NBA 2025-26 Season Play-by-Play Data (pbp2026.csv)

**Aim:** To import and export data across multiple file formats (CSV, Excel, SQL, JSON, Web) and perform exploratory data analysis and missing value cleaning using Pandas.

## Prepare a Working Sample

```python
import pandas as pd
import sqlite3

df_full = pd.read_csv('pbp2026.csv')
df = df_full.sample(300, random_state=42).reset_index(drop=True)

df.to_csv('playbyplay.csv', index=False)
df.to_excel('playbyplay.xlsx', index=False)
df.to_csv('playbyplay.txt', sep='\t', index=False)

print('Files Created Successfully!')
```

## Program 6 - Import SQL Data

```python
conn = sqlite3.connect('playbyplay.db')

df.to_sql('PlayByPlayTable', conn, if_exists='replace', index=False)

sql_df = pd.read_sql('SELECT * FROM PlayByPlayTable', conn)

print('SQL Dataset')
print(sql_df.head())

conn.close()
```

## Missing Value Cleaning

```python
print('Missing Values Before Cleaning')
print(csv_df.isnull().sum())
```

```python
csv_df['x'] = csv_df['x'].fillna(csv_df['x'].mean())
csv_df['y'] = csv_df['y'].fillna(csv_df['y'].mean())
csv_df['dist'] = csv_df['dist'].fillna(csv_df['dist'].mean())
csv_df['team'] = csv_df['team'].fillna(csv_df['team'].mode()[0])
csv_df['player'] = csv_df['player'].fillna(csv_df['player'].mode()[0])

print('Missing Values After Cleaning')
print(csv_df.isnull().sum())
```

```python
csv_df.to_csv('playbyplay_cleaned.csv', index=False)

print('Final cleaned dataset exported as playbyplay_cleaned.csv')
```
