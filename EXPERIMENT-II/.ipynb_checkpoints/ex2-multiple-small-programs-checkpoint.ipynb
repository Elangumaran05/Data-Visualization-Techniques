# Experiment II: Data Import, Export and Cleaning (Multiple Small Programs)

**Dataset:** NBA 2025-26 Season Play-by-Play Data (pbp2026.csv)

Each program below is independent and reloads the dataset on its own, matching the multi-program style of the lab.

## Program: Create Working Sample

```python
import pandas as pd

df_full = pd.read_csv('pbp2026.csv')
sample = df_full.sample(300, random_state=42).reset_index(drop=True)
sample.to_csv('playbyplay.csv', index=False)

print('Dataset Created Successfully')
```

## Program: Combined Filter Condition

```python
import pandas as pd

df = pd.read_csv('playbyplay.csv')

result = df[(df['dist'] > 25) & (df['result'] == 'Made')]

print(result)
```

## Program: Missing Value Cleaning

```python
import pandas as pd

df = pd.read_csv('playbyplay.csv')

print('Before Cleaning')
print(df.isnull().sum())

df['team'] = df['team'].fillna(df['team'].mode()[0])
df['dist'] = df['dist'].fillna(df['dist'].mean())

print('After Cleaning')
print(df.isnull().sum())
```

## Program: Export to Excel and JSON

```python
import pandas as pd

df = pd.read_csv('playbyplay.csv')

df.to_excel('playbyplay.xlsx', index=False)
df.to_json('playbyplay.json')

print('Files Exported Successfully')
```
