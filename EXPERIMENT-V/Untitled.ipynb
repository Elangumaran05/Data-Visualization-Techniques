```python
# ==========================================
# CELL 1: Setup & Data Loading
# ==========================================
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("pbp2026.csv")
print("Dataset loaded successfully for Visualization!")
```

```python
# ==========================================
# CELL 2: Line Chart (Algorithm 1)
# ==========================================
# Filter single game play progression
single_game = df[df["gameid"] == 22500001].dropna(subset=["h_pts"])

plt.figure(figsize=(9, 4.5))
plt.plot(range(len(single_game)), single_game["h_pts"], color="blue", label="Home Score")
plt.title("Home Team Score Progression Over Game Events")
plt.xlabel("Event Index")
plt.ylabel("Points Scored")
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 3: Bar Chart (Algorithm 2)
# ==========================================
# Top 10 teams by total play events recorded
top_teams = df["team"].value_counts().head(10)

plt.figure(figsize=(9, 4.5))
plt.bar(top_teams.index, top_teams.values, color="royalblue", edgecolor="black")
plt.title("Total Plays Recorded per Team (Top 10)")
plt.xlabel("Team Code")
plt.ylabel("Event Count")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 4: Pie Chart (Algorithm 3)
# ==========================================
# Proportion of Made vs Missed Shots
shot_outcomes = df["result"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(shot_outcomes.values, labels=shot_outcomes.index, autopct="%1.1f%%", 
        startangle=140, colors=["#4CAF50", "#FF5722"])
plt.title("Proportion of Made vs. Missed Shots")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 5: Histogram (Algorithm 4)
# ==========================================
# Distribution of Shot Distance
valid_distances = df[df["dist"] > 0]["dist"].dropna()

plt.figure(figsize=(9, 4.5))
plt.hist(valid_distances, bins=25, color="purple", edgecolor="black", alpha=0.7)
plt.title("Shot Distance Frequency Distribution")
plt.xlabel("Shot Distance (ft)")
plt.ylabel("Frequency / Count")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 6: Scatter Plot (Algorithm 5)
# ==========================================
# Court shot locations (X vs Y coordinates)
sample_shots = df[df["type"].isin(["Made Shot", "Missed Shot"])].sample(1500, random_state=42)

plt.figure(figsize=(8, 6))
plt.scatter(sample_shots["x"], sample_shots["y"], alpha=0.4, c="darkcyan", s=20)
plt.title("Scatter Plot of Court Shot Locations (X vs. Y)")
plt.xlabel("Court X Coordinate")
plt.ylabel("Court Y Coordinate")
plt.grid(True, linestyle=":", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 7: Box Plot (Algorithm 6)
# ==========================================
# Detect distance outliers
plt.figure(figsize=(8, 4))
plt.boxplot(valid_distances, vert=False, patch_artist=True,
            boxprops=dict(facecolor="lightblue", color="blue"))
plt.title("Box Plot of Shot Distances (Outlier Detection)")
plt.xlabel("Distance (ft)")
plt.tight_layout()
plt.show()
```

```python
# ==========================================
# CELL 8: Multiple Line Charts (Algorithm 7)
# ==========================================
# Home vs Away score comparison over events
game_data = df[df["gameid"] == 22500001].dropna(subset=["h_pts", "a_pts"])

plt.figure(figsize=(10, 5))
plt.plot(range(len(game_data)), game_data["h_pts"], label="Home Team", color="blue", linewidth=2)
plt.plot(range(len(game_data)), game_data["a_pts"], label="Away Team", color="red", linewidth=2)
plt.title("Home vs Away Score Comparison (Game 22500001)")
plt.xlabel("Event Index")
plt.ylabel("Cumulative Points")
plt.legend(loc="upper left")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```
